```python
from pathlib import Path
import numpy as np
from IPython.display import HTML
from plot import plot_learning_curves, plot_rollout, plot_norms, animate_channels
from utils import relative_norm_error
```

```python
configuration = "vanilla-fno-4-step"
seed = 1
```

```python
results_path = Path("results") / configuration / f"seed_{seed}"
```

```python
import json

with open(results_path / "stats.json", "r") as f:
    stats = json.load(f)
```

```python
print(f"one step train mse = {stats['train']['mse']:.4e}")
print(f"one step val mse   = {stats['val']['mse']:.4e}")
```

```python
print(f"one step train vrmse = {stats['train']['vrmse']:.4f}")
print(f"one step val vrmse   = {stats['val']['vrmse']:.4f}")
```

```python
plot_learning_curves(stats=stats)
```

```python
ref, pred = np.load(results_path / "rollout.npy")
ref.shape, pred.shape
```

```python
ref_bulk, pred_bulk = np.mean(ref, axis=(1, 2)), np.mean(pred, axis=(1, 2))
```

```python
plot_rollout(ref_bulk=ref_bulk, pred_bulk=pred_bulk)
```

```python
ref_norm, pred_norm = relative_norm_error(ref), relative_norm_error(pred)
```

```python
plot_norms(ref_norm=ref_norm, pred_norm=pred_norm)
```

```python
anim_ref = animate_channels(ref)
HTML(anim_ref.to_jshtml())
```

```python
anim_pred = animate_channels(pred)
HTML(anim_pred.to_jshtml())
```

```python
anim_err = animate_channels(ref - pred)
HTML(anim_err.to_jshtml())
```
